**SLICING**


Slicing is used to extract a part of a string using the syntax:      
string[start : stop : step]

start → index from where slicing begins......
stop → index where slicing ends (not included)......
step → number of positions to skip.


```python
s1= "SADAF"
s1= s1[0:3:2]
print(s1)
```

```python
s1= "MY NAME IS SADAF"
s1= s1[-5:-2]    #performing slicing
print(s1)
print(s1[ :6])
s2= "PYTHON"
print(s2[0:10])
s3= "PYTHON"
print(s3[ :len(s3)])
print(s3[-10:-1])
print(s3[ :-1])
print(s3[-6 :])
print(s3[-6 :6])
print(s3[6 :-6])

```


Question:
* First one will give the output
* The second print produces an empty string, so nothing visible is printed.

**MEMBERSHIP:**

Python has two membership operators: in and not in.

```python
s1= "Data"
s2= "Science"
print("a" in s1)
print("b" in s1)
"a" in s1
```
