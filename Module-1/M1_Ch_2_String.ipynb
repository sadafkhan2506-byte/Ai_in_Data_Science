Logical Operator:

Logical operators are used to combine two or more conditions and return True or False.
Python has three logical operators: and, or, and not.

```python
age= int(input("Enter your age:"))
salary= int(input("Enter your salary:"))    #'and' logical operator
if age>= 18 and salary>= 25000:
  print("Eligible for loan")
```

```python
m1= int(input("Enter your marks:"))
if m1>100 or m1<0:      #'or' logical operator
  print("Incorrect marks")
```

```python
its_raining= False
if not its_raining:    #'not' logical operator
  print("Go outside")
```

Else Statement:

The else statement executes a block of code when the if condition is False.
It provides an alternative action when the condition is not satisfied.





```python
no= 100
if no>120:
  print("Eligible")
else:
   print("Number is less than 120")
```

Nested If:

A nested if means using one if statement inside another if statement.
It is used when a second condition needs to be checked only after the first condition is true.

```python
age= 2
has_ID= True
if age>=18:
  if has_ID:
    print("Eligible for vote")
  else:
    print("Not have an ID")
else:
  print("Not Eligible")
```

Elif Statement:

The elif statement is used to check multiple conditions one after another.
It runs the block of the first condition that is True; if none are true, else is executed.

```python
m1=int(input("Enter your marks of HINDI:"))
m2=int(input("Enter your marks of MATH:"))
m3=int(input("Enter your marks of COMP:"))
m4=int(input("Enter your marks of SANS:"))
m5=int(input("Enter your marks of HS:"))
sum= m1+m2+m3+m4+m5
avg= sum/5
if avg>=100 and avg>=90:
  print("Grade A")
elif avg>=90 and avg>=80:
  print("Grade B")
elif avg>=80 and avg>=70:
  print("Grade C")
elif avg>=70 and avg>=60:
  print("Grade D")
else:
  print("Fail")
```

WAP to print leap year:

```python
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```
