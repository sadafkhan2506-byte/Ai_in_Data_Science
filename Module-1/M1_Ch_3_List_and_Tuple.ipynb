### Q. Write a Python program to take 5 names as input from the user and store them in a list. Finally, display the list of names.

```python
my_names = []
for i in range(5):
    name = input(f"Enter name {i+1}: ")
    my_names.append(name)
print("Your list of names:", my_names)
```

### Flatten a List

To flatten a list (i.e., convert a list of lists into a single list), you can use a list comprehension with nested loops.

```python
nested_list = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]

flat_list = [item for sublist in nested_list for item in sublist]

print("Nested List:", nested_list)
print("Flattened List:", flat_list)
```

**Palindrome**

```python
l1 =[1,2,3,2,1]
copy_list = l1.copy()        #Palindrome program using copy() and reverse()
copy_list.reverse()
if l1 == copy_list:
  print(" list is palindrome")
else:
  print("list is not palindrome")
```

```python

lst = [1, 2, 3, 2, 4]
if lst == lst[::-1]:                            #Palindrome using slicing
    print("The list is a Palindrome.")
else:
    print("The list is Not a Palindrome.")
```

Q. Write a Python program to merge two lists into a single list and remove all duplicate elements.

```python

list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]

merged = list1 + list2
result = __builtins__.list(set(merged))

print("Merged List without Duplicates:", result)
```

Q. Write a Python program to create a tuple of numbers from 1 to 8 and extract all the even numbers into another tuple.

```python
numbers = (1, 2, 3, 4, 5, 6, 7, 8)

even_tuple = ()

for i in numbers:
    if i % 2 == 0:
        even_tuple += (i,)

print("Even Tuple:", even_tuple)
```

Q. Write a Python program to separate the even and odd numbers from a given list into two different lists.

```python

numbers = [10, 15, 20, 25, 30, 35]

even = []
odd = []

for i in numbers:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even Numbers:", even)
print("Odd Numbers:", odd)
```

Reference Copy

```python
list1 = [10, 20, 30]

list2 = list1   # Reference Copy

list2.append(40)

print("List1:", list1)
print("List2:", list2)
```

Shallow Copy

```python


list1 = [[1, 2], [3, 4]]

list2 = copy.copy(list1)

list2[0].append(5)

print("Original List:", list1)
print("Shallow Copy:", list2)
```

Deep-Copy

```python


list1 = [[1, 2], [3, 4]]

list2 = copy.deepcopy(list1)

list2[0].append(5)

print("Original List:", list1)
print("Deep Copy:", list2)
```
