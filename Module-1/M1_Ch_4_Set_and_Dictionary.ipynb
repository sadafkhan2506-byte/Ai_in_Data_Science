**Dictionary:**


*A Python dictionary is a mutable data structure that stores information in key-value pairs, where each key must be unique.


*It enables fast data retrieval, updates, and lookups using descriptive keys rather than standard numerical indices.


*Defined using curly braces (e.g., {"name": "Alice", "age": 25}), dictionaries are ideal for representing structured and associative data.





```python
inf= {"name": "Robin","Age":20,"marks": [20,30,25,28],"Status": True}
print(inf)
```

```python
inf= {"name": "Robin","Age":20,"marks": [20,30,25,28],"Status": True, "Empty": " "}
print(inf) #with empty, output will not come
#space dene se execute hojayega
```

```python
inf= {"name1": "Robin","name2":"Rahul", "Age":20,"marks": [20,30,25,28],"Status": True}
print(inf)
```

Nested dictionary:

* A nested dictionary is a dictionary that contains one or more dictionaries inside it as values, enabling the representation of hierarchical data.
 * It is ideal for modeling complex, multi-level structures like JSON data, database records, or configuration settings.
 * You can access inner values by chaining keys together sequentially, such as dict_name['outer_key']['inner_key'].

```python
inf= {"name": "Robin","Age":20,"marks": {"Physics": 20, "Chemistry": 30, "Sanskrit": 35},"Status": True}
print(inf)
```

**Adding tuple to a dictionary:**


* Tuples can be added to a dictionary either as keys (because tuples are immutable and hashable) or as values.
 * To add a tuple as a key, use syntax like my_dict[(1, 2)] = "value", whereas adding it as a value looks like my_dict["key"] = (1, 2).
 * This approach is commonly used for multi-dimensional data, coordinates, or composite keys.

```python
inf= {"name": "Robin","Age":20,"marks": [20,30,25,28],"Status": True, "C":(2,3,4)}
print(inf)
```

SET:

* A Python set is an unordered collection of unique elements, meaning duplicate values are automatically removed.
 * Defined using curly braces {} or set(), sets are mutable, but their individual elements must be immutable and hashable.
 * They are highly efficient for membership testing and performing mathematical set operations like union, intersection, and difference.
