***LOOPS:***


Loops are used to execute a block of code repeatedly until a specific condition is met or for a fixed number of times.
Python mainly has two types of loops: for loop and while loop.

**WHILE LOOP**


A while loop repeatedly executes a block of code as long as the given condition is True.
It is useful when the number of repetitions is not known in advance.

**Factorial**



The factorial of a number is the product of all positive integers from 1 to that number.
It is represented by n!, where n! = n × (n-1) × ... × 1.

```python
n= int(input("Enter a number:"))
fact=1
i=1
while i<=n:
  fact= fact*i      #to print the factorial till 5
  i=i+1
  print("Fact=", fact)
```

Question: Write a Python program using a while loop to search for a specific element (in this case, the number 2) within a given list (n) and print the index where that element is found.

```python
n=[2,4,6,8,10,4,2,12]
idx=0
while idx<len(n):
  if(n[idx]==2):
    print("Element found at index:", idx)
    idx+=1
  else:
    print("Not at index")
  idx+=1
```

Question: Write a Python program using a while loop to search for a specific element (in this case, the number 8) within a given list (n), print the index where the element is found along with a success message, use a break statement to exit the loop once found, and print "Not at index" for the other elements.

```python
n=[2,4,6,8,10,4,2,12]
idx=0
while idx<len(n):
  if(n[idx]==8):
    print("Element found at index:", idx)
    break
  else:
    print("Not at index")
  idx+=1
```

**FOR LOOP**

A for loop is a control flow statement used to iterate over a sequence (such as a list, tuple, string, or range).
It executes a block of code repeatedly for each item in the collection until the entire sequence has been traversed.
This automates element retrieval and makes repetitive programming tasks much cleaner and more efficient.


```python
A=[2,4,6,8,10]
for i in A:
  print(A[i])    #error will be shown here
```

```python
A=[2,4,6,8,10]
for i in A:
  print(i)
```

```python
A=[2,4,6,8,10]
for i in A:
  print(A)
```

```python

#to reverse a number without using reverse or slicing keyword
num = int(input("Enter a number: "))

rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

print("Reversed number is:", rev)
```

```python
#wap to check whether a number is palindrome or not
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

Question: An Armstrong number is a number that is equal to the sum of its own digits, with each digit raised to the power of the total number of digits.

```python
#Armstrong number
print("Armstrong number between 1 and 500:")
num=0
while num<=500:
  num= num+1
  str_num=str(num)
  n= len(str_num)

  arm_num= sum(int(digit)**n for digit in str_num)
  if num==arm_num:
    print(num, arm_num)
```
