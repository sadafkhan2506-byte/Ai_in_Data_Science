File Handling:

* File handling in Python allows programs to create, read, write, and update external files using the built-in open() function.
 * It involves using specific access modes, such as 'r' for reading, 'w' for writing, and 'a' for appending data.
 * Using a with statement (context manager) is the best practice because it automatically and safely closes the file after operations are complete.

writelines():

* The writelines() method in Python is used to write a sequence of strings—such as a list or tuple of lines—directly to a file.
 * It does not automatically insert newline characters (\n), meaning line breaks must be explicitly included within the elements of the sequence.
 * The target file must be opened in a write ('w') or append ('a') mode for the operation to succeed.

```python
f = open("data.txt", "w")

lines = ["Apple\n", "Banana\n", "Mango\n"]   #using the writelines keyword for the next line

f.writelines(lines)
f.close()
```
