###**1**. Series

* A Pandas Series is a one-dimensional labeled array capable of storing data of any type (integers, strings, floats, etc.).
 * It pairs data values with an associated array of labels known as an index, enabling intuitive and fast data access.
 * Serving as a fundamental building block of Pandas, a Series behaves similarly to a single column in a spreadsheet or a SQL table.

```python
import pandas as pd   #importing pandas
marks = [2,4,6,8,10]
ds= pd.Series(marks)
print(ds)
print(type(ds))   #printing the type
```

```python
import pandas as pd
marks = [2,4,6,8,10]       #Customized index
ds= pd.Series(marks, index=["sadaf", "umang", "kani", "riya", "siya"])
print(ds)
```

```python
import pandas as pd
marks = [2,4,6,8,10]       #Customized index
ds= pd.Series(marks, index=["sadaf", "umang", "kani", "riya", "siya"])
print(ds)
```

```python
import pandas as pd
marks = [2,4,6,8,10]       #Customized index
ds= pd.Series(marks, index=["sadaf", "umang", "kani", "riya", "siya"],dtype= "float",name="marksheet record")
print(ds)
```

**1.1 Common Operations**

1.   s.sum()-Calculates and returns the total sum of all numeric  elements in the Pandas Series.
2.   s.mean()- Computes and returns the arithmetic mean (average) of the values in the Series.
3.   s.max()- Returns the highest value present in the Series.
4.   s.min()- Returns the lowest value present in the Series.



```python
print(ds.sum())
print(f"min num in series:", ds.min())
print(f"min num in series:", ds.max())
print(f"min num in series:", ds.mean())
```

Slicing:  Slicing enables the extraction of a subset or range of elements from a sequence using the syntax [start:end:step].

```python
print(f"Access element by Integer number", {ds[0]})   #accessing a value by integer number
```

```python
print(f"Access element by Variable", ds[2])   #accessing the element by variable
```

## **1.3 Vectorization Function:**

* Vectorization allows operations to be applied to entire arrays or series simultaneously, eliminating the need for slow, explicit Python for loops.
 * It executes underlying calculations at C-speeds using libraries like NumPy or Pandas, resulting in massive performance boosts for large datasets.
 * This approach yields cleaner, more concise code and improves efficiency by performing fast, element-wise operations under the hood.

```python
print(ds*3)
print(ds+3)
print(ds-3)
print(ds/3)
print(ds**3)
print(ds%3)
```

**2.1 DataFrame:**




A Pandas DataFrame is a two-dimensional, labeled tabular data structure with rows and columns, closely resembling a spreadsheet or SQL table.

As the primary data structure in Pandas, it is widely used in data science for cleaning, filtering, and analyzing large datasets.

**2.4 Handling Missing Values**

* Handling missing values involves identifying null or NaN entries in a dataset and cleaning them using techniques like imputation or dropping.



```python
import numpy as np
import pandas as pd
Sr={
    "Name": pd.Series(['Sadaf','Riya',np.nan,'Siya','Kani']),
    "Roll_num": pd.Series([10,20,30,40,50]),
    "Marks": pd.Series([10,20,np.nan,40,50]),
    "Subject":pd.Series(["Hindi","Sans","Maths",np.nan,np.nan]),
    "City":pd.Series(["Meerut","Noida","Delhi",np.nan,"Gurgaon"])
}
DF= pd.DataFrame(Sr)
print(DF)
```

* The df.info() method prints a concise summary of a Pandas DataFrame, including its column names, non-null counts, data types, and total memory usage.

```python
DF.info()
```

* The df.isnull() method returns a boolean-valued DataFrame of the same shape, where True indicates a missing value (NaN) and False represents a valid entry.

```python
DF.isnull()
```

* The df.isnull().sum() method counts and returns the total number of missing or NaN values for each column in a Pandas DataFrame.

```python
DF.isnull().sum()
```

* The print(df.isnull().sum().sum()) command computes and returns the grand total of all missing or NaN values present across the entire Pandas DataFrame.

```python
print(DF.isnull().sum().sum())
```

* The dropna() method is used to remove rows or columns that contain missing or NaN values from a Pandas DataFrame or Series.



```python
#remove null values
DF.dropna()
```

```python
DF.dropna(subset=["Marks"])
```

```python
DF["Marks"]=DF["Marks"].fillna(0)
DF
```

* The fillna() method is used to replace missing or NaN values in a Pandas DataFrame or Series with specified values, constants, or methods like forward/backward filling.

```python
DF["Marks"] = DF["Marks"].fillna(DF["Marks"].max())
DF
```
