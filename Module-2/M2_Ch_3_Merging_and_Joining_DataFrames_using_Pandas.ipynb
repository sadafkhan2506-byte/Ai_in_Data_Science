# **MERGE( )- It is column based**

```python
import pandas as pd
df1= pd.DataFrame({'A':[2,3,4,5],'B':["A","B","C","D"]})
df2= pd.DataFrame({'A':[2,4,6,8],'B':["A","C","E","F"]})
# By default, pd.merge performs an 'inner' join and merges on all common column names.
# In this case, 'A' and 'B' are common columns.
# It will find rows where both 'A' and 'B' values match in both DataFrames.
pd.merge(df1,df2)
```

# **JOIN( )- It is index based**

```python
import pandas as pd
V1= pd.DataFrame({'Rollno':[2,3,4,5],'Name':["Sadaf","Umi","Kani","Riya"]})      #Column names should be different otherwise error will come
V2= pd.DataFrame({'ID':[2,3,4,5],'Name2':["Tiya","Mansi","Khushi","Sachi"]})
V1.join(V2)
```

```python
import pandas as pd
V1= pd.DataFrame({'Rollno':[2,3,4,5],'Name':["Sadaf","Umi","Kani","Riya"]},index=['a','b','c','d'])
V2= pd.DataFrame({'ID':[2,3,4,5],'Name2':["Tiya","Mansi","Khushi","Sachi"]})
V1.join(V2)
```

```python
import pandas as pd
V1= pd.DataFrame({'Rollno':[2,3,4,5],'Name':["Sadaf","Umi","Kani","Riya"]},index=[1,2,3,4])
V2= pd.DataFrame({'ID':[2,3,4,5],'Name2':["Tiya","Mansi","Khushi","Sachi"]})
V1.join(V2)
```

```python
import pandas as pd
V1= pd.DataFrame({'Rollno':[2,3,4,5],'Name':["Sadaf","Umi","Kani","Riya"]},index=["A","B","C","D"])
V2= pd.DataFrame({'ID':[20,30],'Name2':["Tiya","Mansi"]},index=["C","D"])
V2.join(V1,how="right")
```

```python
import pandas as pd
V1= pd.DataFrame({'Rollno':[2,3,4,5],'Name':["Sadaf","Umi","Kani","Riya"]},index=["A","B","C","D"])
V2= pd.DataFrame({'Rollno':[20,30],'Name2':["Tiya","Mansi"]},index=["C","D"])    #error will come because col. name is same
# To fix the 'columns overlap but no suffix specified' error, provide suffixes for the overlapping columns.
# 'lsuffix' is for the left DataFrame (V2) and 'rsuffix' is for the right DataFrame (V1).
SI = V2.join(V1,how="right", lsuffix='_v2', rsuffix='_v1')
print(SI)
```

# **CONCAT( )- adds data rather focusing on col. name or index**

```python
import pandas as pd
roll_no=pd.Series([1,2,3,4])
roll_no2=pd.Series([10,20,30,40])
pd.concat([roll_no,roll_no2],axis=0)   #like a line
```

```python
import pandas as pd
roll_no=pd.Series([1,2,3,4])
roll_no2=pd.Series([10,20,30,40])
pd.concat([roll_no,roll_no2],axis=1)    #by using axis=1, adding col. at side of it
```

```python
import pandas as pd
roll_no=pd.Series([1,2,3,4])
roll_no2=pd.Series([10,20,30,40])
pd.concat([roll_no,roll_no2],axis=0,ignore_index=True)   #by using ignore index, it is not repeating the index value again like previous one
```

```python
df1= pd.DataFrame({'A':[1,2,3],'B':["A","B","C"]})
df2= pd.DataFrame({'A':[2,4,6],'B':["A","C","E"]})   #not overriding, as used same col. names
pd.concat([df1,df2],axis=1)
```

```python
df1= pd.DataFrame({'A':[1,2,3],'B':["A","B","C"]})
df2= pd.DataFrame({'D':[2,4,6],'E':["A","C","E"]})
pd.concat([df1,df2],axis=1)
```

```python
df1= pd.DataFrame({'A':[1,2,3],'B':["A","B","C"]})
df2= pd.DataFrame({'D':[2,4,6],'E':["A","C","E"]})
pd.concat([df1,df2],join="inner")  #only iinner/outer can be done
```

```python
df1= pd.DataFrame({'A':[1,2,3],'B':["A","B","C"]})
df2= pd.DataFrame({'D':[2,4,6],'E':["A","C","E"]})
pd.concat([df1,df2],join="right")    #right/left cannot be done, error will come
```

```python
df1= pd.DataFrame({'A':[1,2,3],'B':["A","B","C"]})
df2= pd.DataFrame({'D':[2,4,6],'E':["A","C","E"]})
# When concatenating with 'outer' join (default for columns when axis=0 or rows when axis=1),
#fills NaN for values not present in both.
pd.concat([df1,df2],join="outer")
```

```python
df1= pd.DataFrame({'A':[1,2,3],'B':["A","B","C"]})
df2= pd.DataFrame({'D':[2,4,6],'E':["A","C","E"]})
# Concatenate df1 and df2 along axis=1 (horizontally).

# assigning 'A' to columns from df1 and 'B' to columns from df2.
pd.concat([df1,df2],axis=1,keys=["A","B"])
```

```python
df1= pd.DataFrame({'A':[1,2,3],'B':["A","B","C"]})
df2= pd.DataFrame({'D':[2,4,6],'E':["A","C","E"]})
pd.concat([df1,df2],axis=0,keys=["A","B"])   #keys used to describe the two names to the table
```
