**Task on Line Chart**

```python
# Import the matplotlib library for plotting graphs
import matplotlib.pyplot as plt

# Import the pandas library for data manipulation and analysis
import pandas as pd

# Create a sample dataset containing Month and Profit values
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul"],
    "Profit": [100, 120, 150, 130, 180, 160, 200]
}

# Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# Create a line plot of Profit against Month
# marker="o" adds circular markers at each data point
plt.plot(df["Month"], df["Profit"], marker="o")

# Add a title to the graph
plt.title("Month vs Profit")

# Label the x-axis
plt.xlabel("Month")

# Label the y-axis
plt.ylabel("Profit")

# Rotate the month labels by 45 degrees for better readability
plt.xticks(rotation=45)

# Display the line graph
plt.show()
```

#**3.2 Histogram**

```python
#define dataset
age=[20,21,19,18,20,20,20,21,20,19,17]

#creating histogram chart
plt.hist(age,bins=2,edgecolor="black",color="yellow",alpha=0.3)

#add title and labels
plt.title("Histogram",fontsize=20,color="blue")
plt.xlabel("Age",fontsize=12,color="red")
plt.ylabel("Frequency",fontsize=12,color="red")

plt.grid(axis="y")      #x,y by default full
plt.show()
```

```python
import matplotlib.pyplot as plt      # Import matplotlib for creating graphs
import numpy as np                   # Import numpy for mathematical calculations

age = [20, 19, 8, 19, 16, 20, 21, 23, 28, 19, 23]


plt.hist(age, bins=3, edgecolor='r', color='yellow')

# Add a title to the histogram
plt.title("Age comparison", fontsize=22, color="green")

# Add a label to the x-axis
plt.xlabel("age", fontsize=12, color="red")

# Add horizontal grid lines
plt.grid(axis='y')

# Add a label to the y-axis
plt.ylabel("", fontsize=12, color="red")

# Calculate the median of the age values
median_age = np.median(age)

# Draw a vertical dashed line at the median value
# linestyle='--' makes the line dashed
# linewidth=3.5 controls the thickness of the line
plt.axvline(median_age,color='blue',linestyle='--',linewidth=3.5,label=f'Median = {median_age}')

# Display the legend
plt.legend()

# Display the histogram
plt.show()
```

#**3.3 Bar Chart**

```python
import matplotlib.pyplot as plt           # Import Matplotlib for creating charts

subjects = ["Python", "C", "Java", "DSA"]  # Store the subject names
score = [40, 60, 65, 85]                   # Store the scores for each subject

# Create a bar chart with different colors for each subject
plt.bar(subjects, score, color=["r", "green", "y", "pink"])

plt.title("Subject vs Score", color='r')  # Add the chart title
plt.xlabel("Subjects", color='r')         # Label the x-axis
plt.ylabel("Score", color='r')            # Label the y-axis

plt.grid(axis='y')                        # Add horizontal grid lines for better readability

plt.show()                                # Display the bar chart
```

#**3.4 Pie Chart**

```python
subject = ["Technical", "English", "GK", "Aptitude"]  # Store the names of subjects
score = [40, 60, 65, 85]  # Store the scores of each subject

# Create a pie chart using scores and subject labels
plt.pie(score, labels=subject, autopct="%1.1f%%",
        startangle=45, explode=(0.1, 0, 0, 0))

plt.title("Subject vs Score", color='r')  # Add a red title to the chart

plt.show()  # Display the pie chart
```

**Task on Pie-Chart**

```python
# Import the matplotlib library for creating visualizations
import matplotlib.pyplot as plt

# Group the data by Month and calculate the total Profit for each month
profit_by_month = df.groupby("Month")["Profit"].sum()

# Create a pie chart to show each month's contribution to total profit
plt.pie(
    profit_by_month,              # Profit values
    labels=profit_by_month.index, # Month names as labels
    autopct="%1.1f%%"             # Display percentage with 1 decimal place
)

# Add a title to the pie chart
plt.title("Profit Contribution by Month")

# Display the chart
plt.show()
```

#**3.5 Scatterplot/Bubbleplot Graph**

```python
x1=[2,4,6,8,100]                                   #x-coordinates for the first dataset
y1=[11,13,15,17,19]                                #y-coordinates for the first dataset





x2=[12,14,16,18,90]                               #x-coordinates for the second dataset
y2=[1,3,5,7,9]                                    #y-coordinates for the second dataset


rang=['pink','purple','orange','red','green']     #list of colors for points(will be used for intensity)
size=[100,200,300,400,500]                        #size of each scatter points

plt.scatter(x1,y1,s=size,c=rang,alpha=0.5, label='Dataset 1')            #scatter plot for Dataset 1
plt.scatter(x2,y2,s=size,c=rang,alpha=0.8, label='Dataset 2')            #Scatter plot for Dataset 2


plt.colorbar(label="intensity")     #colorbar to show mapping of color to intensity

plt.legend()                        #show legend for datasets


plt.show()                         #Display the scatter plot
```
